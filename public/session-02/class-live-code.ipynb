## `elif` and `else` — more than two roads

Python checks the branches in order and takes the first true one. Nothing after that runs.


```python
x = 5
if x < 2 :
    print('small')
elif x < 10 :
    print('Medium')
else :
    print('LARGE')
print('All done')
```

## `try` / `except` — when the input is not what you promised

The first cell below fails on purpose. That traceback is the point: `int('Hello Bob')` cannot work, so the program stops. `try`/`except` lets you decide what happens instead of stopping.


```python
astr = 'Hello Bob'
istr = int(astr)
print('First', istr)
```

```python
astr = 'Hello Bob'
try:
    istr = int(astr)
except:
    istr = -1
```

```python
astr = '123'
try:
    istr = int(astr)
except:
    istr = -1

print('Second', istr)
```

## Loops — `while` until something changes, `for` over a collection

`while` needs you to break it. `for` ends on its own when the list runs out.

In class we typed this loop with `input('> ')` and stopped it by typing `done`. In a notebook that pauses *Run all* until someone types, so the cell below feeds the same loop a list of lines instead. The loop is unchanged: `while True`, a test, `break`. The interactive version is in the cell after it — uncomment it if you want to type at it.


```python
typed = ['hello', 'there', 'done', 'this line never prints']
i = 0

while True:
    line = typed[i]
    i = i + 1
    if line == 'done' :
        break
    print(line)
print('Done!')
```

## The "largest so far" pattern

`largest` starts as `None` because we have not seen a number yet. Every loop that finds a maximum, a total, or a count looks like this.


```python
largest = None
print('Before')
for value in [9, 41, 12, 3, 74, 15] :
    if largest is None :
        largest = value
    elif value > largest:
        largest = value
    #print(largest, value)

print('After', largest)
```

## Filtering rows — asking the table a question

`df[df['Sex'] == 'female']` reads as: keep the rows where this is true. Combine conditions with `&` and wrap each one in parentheses.


```python
average_age_female = df[df['Sex'] == 'female']['Age'].mean()
print(f"Average age of female passengers: {average_age_female:.2f}")
```

```python
df[df["Sex"] == "female"].Age.mean()
```

```python
df[df["Age"] < 20].Age.mean()
```

```python
df[(df["Sex"] == "female") & (df["Age"]<20)]
```

```python
# Change the following Python program to extract and count paragraph (p) tags from the retrieved HTML document 
# and display the count of the paragraphs as the output of your program as well as the paragraph texts. 
# Test your program on several small web pages such as
# https://www.suffolk.edu/

# The following Python program is to scrape all anchor tags in a website and print out the href attribute from
# each anchor tag
import urllib.request, urllib.parse, urllib.error
from bs4 import BeautifulSoup
import ssl

# Ignore SSL certificate errors
ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE

url = input('Enter the website address: ')
html = urllib.request.urlopen(url, context=ctx).read()
soup = BeautifulSoup(html, 'html.parser')

# Retrieve all of the anchor tags
tags = soup('a')
for tag in tags:
    print(tag.get('href', None))
```

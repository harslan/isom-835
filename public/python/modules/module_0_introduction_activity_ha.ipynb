## Introduction Activity

In Google Colab Notebook, there are two types of data cells: Code and Text. You can click on +Code button above to crate a code cell. You can click in +Text button to create a text cell. In a code cell, we type our Python to code and run it by clicking on shift-enter. In a text cell, we type our text and render it by clikcing on shift-enter. In Python, we can include single line comments in our code cell by including (#) before our comments. These comments are goign to be ignored by Python interpreter when you run the code cell. You can use also include your comments in triple ''' quotes if they span multiple lines.

Here are some examples:

```
# this is a single line comment in Python. 
# and this is the second line of comment.

4+2
```

```
'''
This is a multiline
comment in Python.
'''
```

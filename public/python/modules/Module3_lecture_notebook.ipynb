# Finding the most occurring word in a text file

```
name = input('Enter file:')
handle = open(name)

# create a frequency dictionary of words from the text file
counts = dict()
for line in handle:
  words = line.split() #turn each string line into a list of words
  for word in words:
    counts[word] = counts.get(word,0) + 1



# loop through the dictionary to find the word with the highest frequency
bigcount = None
bigword = None
for word,count in counts.items():
  if bigcount is None or count > bigcount:
    bigword = word
    bigcount = count

print(bigword, bigcount)
```

# Finding the top 10 most occuring words

```
fhand = open('sampletext.txt')
counts = {}

# a frequency dictionary counts
for line in fhand:
  words = line.split()
  for word in words:
    counts[word] = counts.get(word, 0 ) + 1

# create a list of tuples (frequency, word)
lst = []
for key, val in counts.items():
  newtup = (val, key) 
  lst.append(newtup)

# sort the list
lst = sorted(lst, reverse=True)


for val, key in lst[:10]:
  print(key, val)
```

```
def sum_positional_values(input_string):
    # Initialize the total sum
    total_sum = 0

    # Iterate through each character in the input string
    for char in input_string:
        # Check if the character is a letter
        if char.isalpha():
            # Convert the character to lowercase and calculate its positional value
            # 'a' has value 1, so we subtract ord('a') - 1 to get the correct value
            total_sum += ord(char.lower()) - ord('a') + 1

    return total_sum

# Example usage
input_string = "ABE!!!!"
result = sum_positional_values(input_string)
print(f"The sum of positional values for '{input_string}' is: {result}")
```

```
# Define the sales data dictionary
sales_data = {
    "Boston": [],
    "Cambridge": []
}

# Function to add sales data
def add_sales_data(city, sales):
    if city in sales_data:
        sales_data[city].append(sales)
    else:
        print(f"Error: {city} is not a valid city.")

# Function to get total sales for a city
def get_total_sales(city):
    if city in sales_data:
        return sum(sales_data[city])
    else:
        print(f"Error: {city} is not a valid city.")
        return None

# Function to display sales data
def display_sales_data():
    for city, sales in sales_data.items():
        total_sales = sum(sales)
        print(f"Total sales in {city}: {total_sales}")
        print(f"Individual sales records: {sales}")

# Adding some sample sales data
add_sales_data("Boston", 150)
add_sales_data("Boston", 200)
add_sales_data("Cambridge", 180)
add_sales_data("Cambridge", 220)

# Display the sales data
display_sales_data()

# Get total sales for a specific city
boston_total_sales = get_total_sales("Boston")
cambridge_total_sales = get_total_sales("Cambridge")
print(f"Total sales in Boston: {boston_total_sales}")
print(f"Total sales in Cambridge: {cambridge_total_sales}")
```
